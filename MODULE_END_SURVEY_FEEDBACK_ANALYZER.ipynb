TEXT CLEANING

removes punctuation, extra spaces and capital letters



* created an empty list cleaned_feedbacks=[] to store the cleaned feedback entries
* for feedback in feedback_data['feedback'] the loop takes one feedback at a time from the feedback_list

* for punct in [',' ,'.' ,'!' ,'?']:
    feedback=feedback.replace(punct, '')

    a loop is created .replace() searches for the specified character.

    punct represent the current punctuation mark

    '' is an empty string so the punctuation is removed
*  .split() seperates text wherever there is a white space


*  feedback=''.join(words) this fixes the extra spaces
*  .lower() converts all uppercase letters to lower case


*  cleaned_feedbacks.append(feedback) add the cleaned entry to the new list
*   feedback_data['feedback']=cleaned_feedbacks replace the original feedback list with the cleaned feedback list.









```python
cleaned_feedbacks=[]
for feedback in feedback_data['Feedback']:
  for punct in [',','.','!','?']:
    feedback=feedback.replace(punct, '')

  feedback = ' '.join(feedback.split())
  feedback = feedback.lower()
  cleaned_feedbacks.append(feedback)

feedback_data['Feedback']=cleaned_feedbacks
```

WORD COUNT INSIGHTS

create a function count_word_in_feedbacks using def

*  count=0 we start with zero
*   for feedback in feedback_data['feedback'] takes each cleaned feedback one at a time

*  if word in feedback checks whether the search word is present
*  count=count+1 the counter increase by one for each feedback entry containing the word

*  return count returns the final count to the place where the function was called.




we calculated the word count of good, poor and excellent





```python
def count_word_in_feedbacks(word):
  count=0
  for feedback in feedback_data['Feedback']:
    if word in feedback:
      count=count+1
  return(count)
```

```python
print("Feedbacks containing 'good':",count_word_in_feedbacks("good"))
print("Feedbacks containing 'poor':",count_word_in_feedbacks("poor"))
print("Feedbacks containing 'excellent':",count_word_in_feedbacks("excellent"))
```

Find and display the feedback with the longest comments(in terms of word count)



*   longest feedback='' initialixe an empty string to store the feedback with the highest word count
*  max words=0 initialize maximum word count to zero

*   word count=len(feedback.split()) seperates the feedback into words and len() counts the number of words
*   if word count> max word ckecks whether the current feedback contains more words than the longer feedback found.

*   max word=word count updates the maximum word count when a longer feedback is found
*  longest feedback= feedback stores the current feedback as the longest feedback

*   print(longest feedback) displays the longest feedback
*   print(max words) display the number of words in that feedback









```python
longest_feedback=""
max_words=0
for feedback in feedback_data['Feedback']:
  word_count=len(feedback.split())
  if word_count>max_words:
    max_words=word_count
    longest_feedback=feedback
print("Longest Feedback:",longest_feedback)
print("Number of words:",max_words)
```

Print the list of unique words used across all feedbacks(avoid duplicates)

unique word=set() stores unique values and automatically removes duplicates

*  created a loop to takes each feedback one at a time
*  word=feedback.split() seperates the feedback into individual words

*  for word in words checks each word seperately
*   unique words.add(word) add a word to the set if that word already exists another copy is not added





```python
unique_words= set()
for feedback in feedback_data['Feedback']:
  words=feedback.split()
  for word in words:
    unique_words.add(word)
print("Unique Words:",unique_words)
```

Sort feedback by rating using zip() and sorted() . display the sorted entries


*   zip() combines the rating, name and feedback at the position into one record
*  srt(combined,reverse=true) sorts the record in descending order

*   sorted feedbacks stores the sorted feedbacks
*  print(rating,name,feedback) display each rating along with the corresponding name and feedback





```python
combined=zip(
    feedback_data['Rating'],
    feedback_data['Name'],
    feedback_data['Feedback']
)
sorted_feedbacks=sorted(combined,reverse=True)
for rating,name,feedback in sorted_feedbacks:
  print(rating,name,feedback)
```
