2.String slicing and indexing
a.Print the first character of the string.

```python
print(str3[0]) # python uses the index starting from 0 so index 0 represent first character
```

b. print the last character of the string

```python
print(str3[-1]) # in negative indexing python uses the index starting from -1 so index -1 represent last character
```

c.print the first 5 characters of the string.

```python
print(str3[:5]) # slicing is used to extract characters from the begining upto 5 but not including 5
```

d.print last 11 characters of the string.

```python
print(str3[-11:]) # negative slicing used to get last 11 characters.
```

e.print the string in reverse.

```python
print(str3[::-1]) # the slicing step -1 reversing the string.
```

e.use slicing and print the word'python' from the existing string

```python
print(str3[23:29]) # slicing is used to extract the characters from index 23 to 28 which forms the word python.
```

3. String Methods;
write a python program that takes a srring strM='Python beginner tutorial' and perform the following tasks

```python
strM="Python beginner tutorial"
print(strM)
```

a. convert the sentence to uppercase

```python
print(strM.upper()) # upper() converts all letters to uppercase.
```

b. convert the sentence to lowercase.

```python
print(strM.lower()) # lower() converts all the characters to lowercase.
```

c. use capitalize and return the sentence to the original input form.

```python
print(strM.capitalize()) # capitalize() converts first character of the string to uppercase and remaining characters to lowercase
```

d. count the total number of occurrences of character 't' in the string.

```python
print(strM.count('t')) # count() used to find how many times the character t occurs in the string.
```

e. replace all occurrences of 'Python' with 'Machine Learning' in the input string.

```python
print(strM.replace('Python', 'Machine Learning')) # the replace() mwthod is used to replace python with machine learning
```

b. repeat the elements of t_combine 3 times

```python
print(t_combine * 3) # *operator is used to repeat the tuple elements 3 times.
```

c. access the 3rd element from t_combine

```python
print(t_combine[2]) # python indexing starts from 0 index 2 represents the 3rd element.
```

d. access the first 3 elements from t_combine

```python
print(t_combine[:3]) # slicing is used to extract first 3 characters.
```
