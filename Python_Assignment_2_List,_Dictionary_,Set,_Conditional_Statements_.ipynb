```python
name_list.append('Yazhini') # append() is used to add the name in the list.
print(name_list)
```

b.insert the element 30 at index 2 in age_list

```python
age_list.insert(2,30) # to insert the age in the corresponding index position we used insert()
print(age_list)
```

```python
name_list.remove('Yazhini') # remove() is used to remove the name
print(name_list)
```

d. pop the last element from age_list

```python
age_list.pop() # pop() without giving index number removes the element at last index position.
print(age_list)
```

e. extend the age list with additional ages[29,30,26]

```python
a=[29,30,26]
age_list.extend(a) # extend() is used to add multiple element to the end of existing list.
print(age_list)
```

```python
age_list.sort(reverse=True) # to print age_lisit in descending order use sort(reverse=True)
print(age_list)
```

g. find max age,min age and sum of all ages from age_list

```python
print(max(age_list)) # to find max age we used max()
print(min(age_list)) # to find min age we used min()
print(sum(age_list)) # to find sum we use sum()
```

c. create two sets

set1 with values{1,3,5,7,9}

set2 with values {2,3,5,8,10}

```python
print(set1.union(set2)) # union() used to get union which gives all values from set1 and set2 removing duplicates if exist
print(set1.intersection(set2)) # intersection() is used to get common values in set1 and set2
```

OPERATORS AND CONDITIONAL STATEMENTS

```python
score=float(input('Enter your score(0 to 10): '))
if score >7:
  print('Above Average : Excellent performance')
elif 4 <= score <= 7:
  print('Average : Good effort keep practising')
else:
  print('Below Average: Need to improve your performance')
```

this program takes a score as input and uses if,elif and else statements to identify students performance category based on the score
