```python
import random
secret_number = random.randint(1,10)
attempts=3

while attempts>0:
  guess=int(input('Guess the number between 1 and 10:'))
  if guess<1 or guess>10:
    print('Your guess is out of range. Guess a number between 1 and 10.')
    attempts-=1
    continue
  attempts-=1

  if guess > secret_number:
    print('Too high. Try again.')
  elif guess< secret_number:
    print('Too low. Try again.')
  else:
    print('Congratulations! You guessed the correct number.')
    break
else:
  print('Better luck next time!')
  print('The correct number was:', secret_number)
```

BMI CALCULATOR

We define function called calculate_bmi. weight and height are parameters that receive the values needed to calculate BMI.

bmi=weight/(height**2) is the formula used to calculate bmi

bmi=calculate_bmi(weight,height)

we call calculate_bmi() function and pass the users weight and height to it. the returned bmi is stored in the bmi variable.

print('your bmiis:', round(bmi,2))

this display the calculated bmi anf format the result to two decimal points.

```python
def calculate_bmi(weight,height):
  bmi=weight / (height**2)
  return bmi
weight=float(input('Enter your weight in kilograms:'))
height=float(input('Enter your height in meters:'))

bmi=calculate_bmi(weight,height)
print('Your BMI is:',round(bmi,2))
```
